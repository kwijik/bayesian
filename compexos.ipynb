```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
```

```python
n = 100
data = np.random.normal(loc=0, scale=1, size=n)
sorted = np.sort(data)
```

```python
sorted
```

```python
Fn = np.arange(1, n+1) / n
```

```python
alpha = 0.05
```

```python
epsilon = np.sqrt(np.log(2/alpha) / (2*n))
```

```python
lower = np.maximum(Fn - epsilon, 0)
upper = np.minimum(Fn + epsilon, 1)
```

```python
# true CDF
x_grid = np.linspace(-3, 3, n)
true_cdf = norm.cdf(x_grid)

# plot
plt.step(sorted, Fn, where="post", label="ECDF")
plt.step(sorted, lower, where="post", linestyle="--", label="Lower band")
plt.step(sorted, upper, where="post", linestyle="--", label="Upper band")

plt.plot(x_grid, true_cdf, label="True CDF")

plt.legend()
plt.title("95% Confidence Band for CDF (DKW)")
plt.show()
```
